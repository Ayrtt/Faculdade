# Exercício 1 - Dataset Digits do sklearn

Acesso: https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_digits.html#sklearn.datasets.load_digits

1) Importar o pacote "sklearn.datasets" e o "load_digits"

```python
import sklearn.datasets
import numpy as np
import pandas as pd
from sklearn import metrics
from sklearn.datasets import load_digits
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

2) Carregar o dataset através do método: load_digits()
3) Observe as keys do dataset usando o método "keys"

```python
digits = load_digits()
digits.keys()
```

4) A chave "data" são as features e a chave "target" é o y. Separe os dados em 2 variáveis diferentes
5) Verificar a dimensionalidade das features através da variável shape

```python
y = digits.target
X = digits.data
print(y.shape)
print(X.shape)
```

6) Separe o conjunto de dados em treinamento e teste usando o método: "train_test_split"

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y) # 80% treino e 20% teste
```

7) Treinar MLP (2 topologias diferentes)

```python
mlpTanh = MLPClassifier(hidden_layer_sizes=(20,10), activation='tanh',max_iter=2000).fit(X_train, y_train)
```

```python
mlpIdentity = MLPClassifier(hidden_layer_sizes=(20,10), activation='identity',max_iter=2000).fit(X_train, y_train)
```

8) Treinar Árvore de Decisão (com Entropia e Gini)

```python
treeEntropy = DecisionTreeClassifier(criterion="entropy").fit(X_train, y_train)
```

9) Treinar Árvore de Decisão com Max_Length = 2

```python
treeGiniD2 = DecisionTreeClassifier(max_depth = 2).fit(X_train, y_train)
```

10) Treinar KNN (com duas características diferentes)

```python
knnEuclidean = KNeighborsClassifier(n_neighbors=7, metric='euclidean', algorithm='brute').fit(X_train, y_train)
```

```python
knnBrute = KNeighborsClassifier(n_neighbors=7, algorithm='brute').fit(X_train, y_train)
```

11) Mostrar a taxa de acerto de todos os modelos

```python
def Acc(model):
  result = model.predict(X_test)
  acc = metrics.accuracy_score(result, y_test)
  show = round(acc * 100)
  print("{}%".format(show))
```

```python
Acc(mlpTanh)
Acc(mlpIdentity)
Acc(treeEntropy)
Acc(treeGiniD2)
Acc(knnEuclidean)
Acc(knnBrute)
```
