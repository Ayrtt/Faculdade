```python
df = X3.values
min_max_scaler = MinMaxScaler()
x_scaled = min_max_scaler.fit_transform(df)
X3 = pd.DataFrame(x_scaled)
```

5) Separar os dados em treinamento e teste

```python
X3_train, X3_test, y3_train, y3_test = train_test_split(X3, y3, test_size=0.2)
```

```python
logisticR_ex3 = LogisticRegression(random_state=0).fit(X3_train, y3_train).predict(X3_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
tree_ex3 = DecisionTreeClassifier().fit(X3_train, y3_train).predict(X3_test)
```

8) Treinar um KNN para classificação

```python
knn_ex3 = KNeighborsClassifier().fit(X3_train, y3_train).predict(X3_test)
```

9) Apresentar os resultados de acurárica em percentual
